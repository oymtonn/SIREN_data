```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
# count rates of all muons vs coincidence muons (single dataset)

master = pd.read_csv("final_master_data.csv")
slave = pd.read_csv("final_slave_data.csv")

plt.plot(master["time_stamp"], master["count"], label="All Detections")
plt.plot(slave["time_stamp"], slave["count"], label="Coincidence Detections")
plt.title("Muon Count Rates Over Time")
plt.xlabel("Time Stamp")
plt.ylabel("Count")
plt.legend()
```

```python
slave = pd.read_csv("final_slave_data.csv")
master = pd.read_csv("final_master_data.csv")

bins = np.linspace(
    # use min of two datasets avoid starting where theres no data
    min(slave["adc"].min(), master["adc"].min()),
    # use 99 percentile to avoid outliers dominating the plot
    max(slave["adc"].quantile(0.99), master["adc"].quantile(0.99)),
    # 40 evenly spaced bins
    40
)


plt.hist(master["adc"], bins=bins, alpha=0.5, label="All Detections", density=True)
plt.hist(slave["adc"], bins=bins, alpha=0.5, label="Coincident Detections", density=True)
plt.ylabel("Normalized Number of Events")
plt.xlabel("ADC Value")
plt.title("ADC Value Distribution")
plt.legend()

# shows that the coincident detections have a higher ADC value on average, which is expected since they are more likely to be real muon events rather than noise
```

```python
slave = pd.read_csv("final_slave_data.csv")
master = pd.read_csv("final_master_data.csv")


bins = np.linspace(
    min(slave["sipm_voltage"].min(), master["sipm_voltage"].min()),
    max(slave["sipm_voltage"].quantile(0.99), master["sipm_voltage"].quantile(0.99)),
    40
)

plt.hist(master["sipm_voltage"], bins=bins, alpha=0.5, label="All Detections", density=True)
plt.hist(slave["sipm_voltage"], bins=bins, alpha=0.5, label="Coincident Detections", density=True)
plt.xlabel("SiPM Voltage (mV)")
plt.ylabel("Normalized Number of Events")
plt.title("SiPM Voltage Distribution")
plt.legend()
```
